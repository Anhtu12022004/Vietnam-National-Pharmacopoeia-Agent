```python
def compare_files_ignore_hash(file1_path, file2_path):
    try:
        # Mở và đọc nội dung 2 file
        with open(file1_path, 'r', encoding='utf-8') as f1, \
             open(file2_path, 'r', encoding='utf-8') as f2:
            
            lines1 = f1.readlines()
            lines2 = f2.readlines()
            
        # Lấy số dòng của file dài hơn để tránh lỗi index
        max_len = max(len(lines1), len(lines2))
        
        diff_found = False
        
        for i in range(max_len):
            # Lấy nội dung dòng hiện tại, nếu file ngắn hơn đã hết dòng thì gán chuỗi rỗng
            l1 = lines1[i] if i < len(lines1) else ""
            l2 = lines2[i] if i < len(lines2) else ""
            
            # Loại bỏ dấu thăng (#) và khoảng trắng thừa (bao gồm ký tự xuống dòng)
            clean_l1 = l1.replace('#', '').strip()
            clean_l2 = l2.replace('#', '').strip()
            
            # So sánh 2 dòng sau khi đã làm sạch
            if clean_l1 != clean_l2:
                diff_found = True
                print(f"Dòng {i + 1} có 2 dòng text khác nhau:")
                print(f" + File 1: {l1.strip()}")
                print(f" + File 2: {l2.strip()}")
                print("-" * 60)
                
        if not diff_found:
            print("Hai file hoàn toàn giống nhau (sau khi bỏ qua dấu #).")
                
    except FileNotFoundError as e:
        print(f"Lỗi: Không tìm thấy file - {e}")

# Chạy thử với 2 file bạn đã tải lên
if __name__ == "__main__":
    file_1 = 'page_711_1029.md'
    file_2 = 'page_711_1029_normalized.md'
    compare_files_ignore_hash(file_1, file_2)
```
